**Task 2.1**

Write a function `task2_1(filename)` that:

- takes a string `filename` which represents the name of a text file
- reads in the contents of the text file
- returns the content as a list of integers. **[3]**

Call your function `task2_1` with the file `TEN.txt`, printing the returned list and its length, using the following statements:

```
result = task2_1('TEN.txt')
print(result)
print(len(result))
```

**[2]**


````python
# Task 2.1
def task2_1(filename):
    num = []
    file = open(filename, 'r')
    for line in file:
        line = line.strip()
        num.append(int(line))
    file.close()
    return num
#main
result = task2_1('TEN.txt')
print(result)
print(len(result))
````

**Task 2.2**

One method of sorting is the insertion sort.

Write a function `task2_2(list_of_integers)` that:

- takes a list of integers
- implements an insertion sort algorithm
- returns the sorted list of integers. **[5]**

Call your function `task2_2` with the contents of the file `TEN.txt`, printing the returned list, for example, using the following statement:

`print(task2_2(task2_1('TEN.txt')))` **[1]**


````python
# Task 2.2
def task2_2(list_of_integers):
    ar = list_of_integers
    for i in range(1,len(ar)): # iteration starts from second item. eg. i = 1,2,3,4,5
        current = ar[i]
        j = i-1 #j keeps track of the index of the last item in the sorted sub array index in each iteration

        #finding the 1st occurance in sorted sublist an item > current
        #j keeps track of the items in the sorted sublist
        while j >= 0 and ar[j] > current: #j >= 0 to ensure not all items are in sorted sublist 
            #ar[j] > current to ensure num in ar[j] > comparing int and indicates that position to insert current not yet found
            ar[j+1] = ar[j]#move ar[j] item one position backward
            j -= 1 #move index of sorted sublist to the next item to compare
        ar[j+1] = current #copy value of current in the correct position in the sorted sublist
    return ar
#main
print(task2_2(task2_1('TEN.txt')))
````

**Task 2.3**

Another method of sorting is the quicksort.

Write a function `task2_3(list_of_integers)` that:

- takes a list of integers
- implements a quicksort algorithm
- returns the sorted list of integers. **[7]**

Call your function `task2_3` with the contents of the file `TEN.txt`, printing the returned list, for example, using the following statement:

`print(task2_3(task2_1('TEN.txt')))` **[1]**


````python
# Task 2.3
def task2_3(list_of_integers):
    return quicksort(list_of_integers, 0, len(list_of_integers) - 1)

def partition(ar,low,high):
    pivot = ar[high] #selecting last item as pivot
    i = low - 1
    j = low
    while j <= high:
        if ar[j]< pivot:
            i += 1
            ar[j],ar[i] = ar[i],ar[j] #swap values of ar[j] and ar[i]
        j += 1 #increment index and move to the next item in array/sub-array
    i += 1
    ar[i],ar[high] = ar[high],ar[i]
    return i

def quicksort(ar,low,high):  #low is start index, high is end index
    if low < high:
        #pi is partitioning index, ar[p] is now at right place
        pi = partition(ar,low,high)
        #Separately sort elements before partition and after partition
        quicksort(ar,low,pi - 1)#calls itself
        quicksort(ar,pi+1,high) #calls itself
    return ar

    
#main
print(task2_3(task2_1('TEN.txt')))
````
