````python
# Task 3.1
class node():
    def __init__(self):
        self.data = 0
        self.next = -1
    
    def getdata(self):
        return self.data
    
    def setdata(self, data):
        self.data = data
    
    def getnext(self):
        return self.next
    
    def setnext(self, Next):
        self.next = Next


class LinkedList():
    def __init__(self):
        self.head = -1
        self.nextfree = 0
        self.ar = []

    def insert(self, integer_value):
        self.ar.append(node())
        if self.head == -1:
            self.ar[self.nextfree].data = integer_value
            self.head = self.nextfree
            self.nextfree = self.ar[self.nextfree].next
            self.ar[self.head].next = -1
        else:
            new_index = len(self.ar) - 1
            self.ar[new_index].data = integer_value
            self.ar[new_index].next = self.head
            self.head = new_index
    
    def delete(self, integer_value):
        if self.head == -1: #linked list empty
            return None
        prev = -1
        cur = self.head
        while cur != -1:
            if self.ar[cur].data == integer_value: #if item is found
                break
            prev = cur
            cur = self.ar[cur].next
        if cur == -1:  # 1. check if not found
            print(integer_value, "not found. Nothing deleted!")

        elif prev == -1:  # 2. Case: Deleting the first item
            print(self.ar[cur].data, "deleted")
            temp = self.ar[self.head].next
            self.ar[self.head].next = self.nextfree
            self.nextfree = self.head
            self.head = temp

        elif self.ar[cur].next == -1:  # 3. Case: Deleting the last item
            print(self.ar[cur].data, "deleted")
            self.ar[cur].next = self.nextfree
            self.nextfree = cur
            self.ar[prev].next = -1

        else:  # 4. Case: Deleting a middle item (Your new code, cleaned up)
            print(self.ar[cur].data, "deleted")
            self.ar[prev].next = self.ar[cur].next  # Bypass current node
            self.ar[cur].next = self.nextfree       # Link deleted node to free list
            self.nextfree = cur                     # Move free list head to deleted node

    def count(self):
        if self.head == -1:
            return 0
        count = 0
        for i in self.ar:
            count += 1
        return count
    
    def to_String(self):
        array = []
        cur = self.head
        while cur != -1:
            array.append(self.ar[cur].data)
            cur = self.ar[cur].next
        print(array)

lst = LinkedList()
print("inserting")
lst.insert(11)
lst.insert(2)
lst.insert(7)
lst.insert(4)
print("printing")
lst.to_String()
lst.delete(2)
lst.delete(4)
lst.delete(11)
lst.to_String()
````
